```python
import seaborn as sns
import matplotlib.pyplot as plt

# Loading dataset
data = sns.load_dataset("tips")

# Box plot
sns.boxplot(data=data.groupby("day")["total_bill"].median()), x="day", y="total_bill", hue="sex")

# Labels and title
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.title("Total Bill Distribution by Day and Sex")

plt.show()
# Find median bill for each day
print(data.groupby("day")["total_bill"].median())
print(data.groupby("day")["total_bill"].median().idxmax())
```
