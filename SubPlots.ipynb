```python
#  subplots
#  -- Subplots allow you to display multiple plots 
#- same data
# - differe data set comparison
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [2, 3, 5, 7, 11]
y2 = [1, 4, 2, 5, 3]

# 1 row, 2 columns , first subplot
plt.subplot(1,2,1) # (rows, column, index)
plt.plot(x,y1,color = 'blue',marker = 'o')
plt.title("First Subplot")
plt.xlabel("X")
plt.ylabel("Y")

# 1 row, 2 columns , 2nd subplot
plt.subplot(1,2,2) # (rows, column, index)
plt.plot(x,y1,color = 'red',marker = 's')
plt.title("Second Subplot")
plt.xlabel("X")
plt.ylabel("Y2")

plt.tight_layout()
plt.show()
```

```python
# plt.subplot(nrows, ncols, index) selects the current axes.
# Index counts rows wise from top left to bottom right. (Row major order).
```
