```python
# Matplotlib..
#  it is a python library used for:

# --creating static, animated & interactive plots.
# --low-level control over plot appearence.
# --serving as the foundation for other libraries.



# Basic Plot Structure

import matplotlib.pyplot as plt

X = [1,2,3,4]
Y = [5,6,7,8]
plt.plot(X,Y)  # line plot

```

```python
# Line Chart..
# - A line chart is a type of chart used to display data points connected by straight lines,
# typically to show trends or changes over a continuous variable, such as time and distance.


# single line plot.

# oscar winner movies
oscur_movie = [
    "the dark knight",
    "the hurt locker",
    "the king's speech",
    "the artist",
    "argo"
]
years = [2008,2009,2010,2011,2012]
oscur_revenue = [1005,170,427,133,232]

# plot
plt.plot(years, oscur_revenue)
plt.title("Oscar Movies Revenue in each Year")
plt.xlabel("years")
plt.ylabel("Revenue (in $m)")
```

```python
# multiple lines on the same plot (line chart)

non_oscar_movies = ["slumgdog millionaire","avatar","inception","hugo","lincoln"
]
non_oscar_revenue = [378,2788,829,185,275]

plt.plot(years, oscur_revenue, "o--r",label = "Oscar Movies")
plt.plot(years, non_oscar_revenue, "<-y",label = "Non-Oscar Movies")
plt.title("Oscar Movies vs Non-Oscar Movies Revenue")

plt.xlabel("years")
plt.ylabel("Revenue (in $m)")
plt.legend()

# to print all style in the chart
# plt.style.available
# plt.style.use("dark_background")
plt.style.use("default")
plt.show()
```
