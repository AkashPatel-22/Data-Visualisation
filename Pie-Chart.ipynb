```python
# Pie Chart
# -- A pie chart is a circular chart divided into slices to shoe relative proportions of a whole.
#  -- Each slice = a category
#  -- Size of slice = value of that category relative to the total.

# Customaizations:

# 1. autopct = "%1.1f%%" - show percenatge of each slice.
# 2. colors = add custom colours.
# 3. wedgeprops = define (edgecolor, linewidth,line style, aplha)
# 4. explode =  highlights a slice by pulling it out.
# 5. startangle = rotate the chart.
# 6. shadow = adds 3D efffect or makes the chart circular.


import matplotlib.pyplot as plt

expenses = ["Salaries","Rent","Marketing","R&D","Miscellaneous"]
amounts = [500,150,120,100,50]

explode = [0,0,0,0.2,0]

plt.style.use("def")
plt.pie(amounts,labels=expenses,autopct="%1.1f%%",
       wedgeprops = {
           "edgecolor":"black",
           "linewidth":1,
           "linestyle": "--"
       },
        explode = explode,
        startangle = 90
       )
plt.show()
```
