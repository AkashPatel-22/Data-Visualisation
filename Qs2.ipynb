```python
import matplotlib.pyplot as plt
import seaborn as sns

# loading data
penguins = sns.load_dataset("penguins")
print(penguins.columns)
sns.scatterplot(
    data=penguins,
    x="body_mass_g",
    y="flipper_length_mm",
    hue="species"
)

plt.show()
```
