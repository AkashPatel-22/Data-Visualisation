```python
# Q1.For the following dataset generated using NumPy:

# np.random.normal(loc=70, scale=10, size=200)

# --Plot a histogram of scores
# --Draw a vertical line for the mean & try to interpret the results.
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate data
scores = np.random.normal(loc=70, scale=10, size=200)

# Calculate mean
mean_score = np.mean(scores)

# Plot histogram
plt.figure(figsize=(8,5))
plt.hist(scores, bins=15, edgecolor='black', alpha=0.7)

# Draw vertical line for mean
plt.axvline(mean_score, color='red', linestyle='--',
            linewidth=2, label=f'Mean = {mean_score:.2f}')

# Labels and title
plt.xlabel('Scores')
plt.ylabel('Frequency')
plt.title('Histogram of Scores')
plt.legend()

plt.show()
```
