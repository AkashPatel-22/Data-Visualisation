```python
# Bar Chart..
# -- A bar chart is graph that displays categorical data using rectangular bars.
# Each bar's height represents a value.

# oscar winner movies
import matplotlib.pyplot as plt

oscur_movie = [
    "the dark knight",
    "the hurt locker",
    "the king's speech",
    "the artist",
    "argo"
]
years = [2008,2009,2010,2011,2012]
oscur_revenue = [1005,170,427,133,232]

plt.bar(years,oscur_revenue)
plt.title("Revenue in each year for oscar movies")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

# plt.text - hum apne bar k liye kuch label create kr skt h
# plt.text(years[0],oscur_revenue[0]+15,str(oscur_revenue[0]),ha = "center")

for i in range(len(years)):
    plt.text(years[i],oscur_revenue[i]+10,str(oscur_revenue[i]),ha = "center")
```

```python

# Comparing Multiple Bar Charts

import numpy as np

oscur_movie = [
    "the dark knight",
    "the hurt locker",
    "the king's speech",
    "the artist",
    "argo"
]
years = [2008,2009,2010,2011,2012]
oscur_revenue = [1005,170,427,133,232]

non_oscar_movies = ["slumgdog millionaire","avatar","inception","hugo","lincoln"]
non_oscar_revenue = [378,2788,829,185,275]

years = [2008,2009,2010,2011,2012]

x = np.arange(len(years))
width = 0.4
plt.bar(x-width/2,oscur_revenue,width,label = "Oscar Movies") # shifting values to left
plt.bar(x+width/2,non_oscar_revenue,width,label = "Non-Oscar Movies") # shifting values to right

# width = 0.25
# plt.bar(x-width,oscur_revenue,width,label = "Oscar Movies") 
# plt.bar(x,oscur_revenue,width,label = "Oscar Movies") 
# plt.bar(x+width,non_oscar_revenue,width,label = "Non-Oscar Movies") 

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("years")
plt.ylabel("Revenue (in $m)")
```
