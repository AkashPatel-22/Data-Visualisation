```python
# SeaBorn
# -- seabron is a high level pyhton data visualization libarary built on
# top of matplotlib

import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme()
sns.get_dataset_names()

tips = sns.load_dataset("tips")
# print(type(tips))
# print(tips.head)

sns.relplot(data=tips, x="total_bill", y="tip",hue="smoker",size="size",style="smoker")

# for line plot.
# sns.relplot(data=tips, x="total_bill", y="tip",kind ="line")
plt.show()
```

```python
# Relational Plots.

# --Relational plots are plots that show the relatioship between two(or more)
# variables typically numerical x and y variables.
# Inludes = 'sacatter & line plots'.

# Scatter plot.
tips = sns.load_dataset("tips")
sns.scatterplot(data = tips,x="total_bill",y="tip",hue="time")
```

```python
# line plot
flights =  sns.load_dataset("flights")
sns.lineplot(data=flights,x="year", y = "passengers")
```

```python
#  Categorical Plots.
# -- categorical plots are used to vusualize data where one variable is categorical
# (discrete) and the other is usually numerical.
 # Includes = 'BAR' & 'BOX' plots

# bar plots.
sns.barplot(data=tips,x="day",y="tip",hue="sex")
```

```python
# box plots.
sns.boxplot(data=tips,x="day",y="tip")
```

```python
# Distribution Plots
# -- distribution plots are used to visualize the distribution of single numerical variables.

# histogrm
tips = sns.load_dataset("tips")
sns.histplot(data=tips,x="total_bill",bins=10,kde=False)
plt.title("Histogram total bill")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt


# Matrix Plots
# -- matric=x plots are used to visualize 2D data arranged in a matrix or table,
# such as correlation matrices, confusion matrices or pivot tables.

tips = sns.load_dataset("tips")

corr = tips.select_dtypes(include='number').corr()# correlation btw numerical columns

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
