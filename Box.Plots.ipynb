```python
# BOX Plot
# -- A box plot is a statistical visualization that summarizes the distribution of each dataset using five key numbers;

# - minimum (lowest non-outlier).
# - first Quartile(Q1) - 25th percentile.
# - Median (Q2) - 50th percentille.
# - Third Quartile (Q3) - 75th percentile.
# - maximum (highest non-outlier).

# It can also show 'outliers'

import matplotlib.pyplot as plt

data = [7,8,5,6,9,4,10,12,15]
plt.boxplot(data) 
plt.grid(True)
```

```python
# multiple datasets.
import numpy as np
grp1 = np.random.normal(50,10,100)
grp2 = np.random.normal(60,15,100)

plt.boxplot([grp1,grp2], tick_labels=['group1','group2'],showmeans=True)
plt.grid(True)
plt.show()
```

```python
# horizontal box plots

data = [7,8,5,6,9,4,10,12,15]
plt.boxplot(data,vert=False,showmeans=True,whis=2.0) 
plt.grid(True)
```
