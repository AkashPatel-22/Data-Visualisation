```python
# Stack Plots.
# -- A stack plot is atype of plot where data series are stacked on the top of each other.
# Each "layer" shows the contribution of one category , adn the top line shows the cumulative total.

import matplotlib.pyplot as plt

days = ["mon","tues","wed","thu","fri","sat","sun"]
direct = [50,60,70,80,90,100,110]
organic = [30,40,50,55,60,70,80]
social = [20,25,30,35,40,50,60]

# plt.pie([direct[0],organic[0],social[0]],labels = ["Direct","Organic","Social"],autopct="%1.1f%%")
plt.stackplot(days, direct,organic,social,labels = ["Direct","Organic","Social"])
plt.title("Marketing data for the Week")
plt.xlabel("Days")
plt.ylabel("# oof customers")
plt.legend()
```

```python
# When to use stack plots?

# Stack plots are used to visualize the contribution of each category to the whole over time.
# - show cumulative data over time.
# -campare multiple components.
# -Highlight trends in parts and whole.
```
